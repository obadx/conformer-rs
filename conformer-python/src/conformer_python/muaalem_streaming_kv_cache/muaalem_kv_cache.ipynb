```python
from transformers import PretrainedConfig

class MyStreamingConfig(PretrainedConfig):
    model_type = "streaming_kv"

    def __init__(self, hidden_size=1024, num_hidden_layers=24, 
                 num_attention_heads=16,
                 chunk_size=25,
                 max_context_frames=250,
                 use_sliding_window = False, #limit how far back attention can look
                 sliding_window_left=25,
                 sliding_window_right=5,
                 **kwargs):
        super().__init__(**kwargs)
        self.hidden_size = hidden_size
        self.num_hidden_layers = num_hidden_layers
        self.num_attention_heads = num_attention_heads
        self.chunk_size = chunk_size
        self.max_context_frames = max_context_frames
        self.use_sliding_window = use_sliding_window    
        self.sliding_window_left = sliding_window_left
        self.sliding_window_right = sliding_window_right
        
```

```python
my_config = MyStreamingConfig()
print(my_config)
```
